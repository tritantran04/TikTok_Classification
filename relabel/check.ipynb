**Check is there any problems in relabling**

```python
import os 
import json
mapping_file = {}
# Compare the paths in path_storage['video_path'] to the relabeled paths
label_list = list(path_storage['class_name'].unique())
for idx, sample in path_storage.iterrows():
    label = sample['class_name']
    real_path = f"../datasets/{sample['video_path']}"
    if not os.path.exists(real_path):
        print(f'The path {real_path} is false')
        
        # Check is it moved into another folder
        for l in label_list:
            if l != label:
                check_path = "/".join(real_path.split('/')[:-2]) + "/" + str(l) + "/" + sample['video_name']
                if os.path.exists(check_path):
                    print(f"!!! It is moved to  {check_path}!!!")
                    if mapping_file.get(real_path, None):
                        print('This video exists in more than one folder')
                    mapping_file[sample['video_path']] = '/'.join(check_path.split('/')[2:])
                    
with open('relabel.json', 'w', encoding='utf-8') as f:
    json.dump(mapping_file, f, ensure_ascii=False, indent=4)
```
